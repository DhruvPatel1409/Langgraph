```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
def create_outline(state: BlogState) -> BlogState:

    # fetch the title
    title = state["title"]

    # call LLM to generate outline
    prompt = f"""
    Generate a detailed outline for a blog post titled "{title}".
    The outline should be a list of 3-5 key points.
    """
    outline = model.invoke(prompt).content

    # update the state
    state["outline"] = outline

    return state
    
```

```python
def create_blog(state: BlogState) -> BlogState:

    # fetch the title and outline
    title = state["title"]
    outline = state["outline"]

    # call LLM to generate blog content
    prompt = f"""
    Write a detailed blog post titled "{title}" with the following outline:
    {outline}
    """
    content = model.invoke(prompt).content

    # update the state
    state["content"] = content

    return state
```

```python
graph = StateGraph(BlogState)

# nodes

graph.add_node("create_outline", create_outline)
graph.add_node("create_blog", create_blog)

# edges
graph.add_edge(START, "create_outline")
graph.add_edge("create_outline", "create_blog")
graph.add_edge("create_blog", END)

# compile the graph
workflow = graph.compile()
```

```python
# execute the graph
input_data = {"title": "Rise of AI in India"}
result = workflow.invoke(input_data)
print(result)
```
